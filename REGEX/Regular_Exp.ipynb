#Character Set[]

```python
import re
```

```python
txt = "The train is running"

x = re.findall(f"[a-m]",txt)

print(x)
```

#Escape Sequence

```python
#for print digit
txt = 'Hi This is Rayhan, My age is 24'

x = re.findall(r"\d+",txt)

print(x)
```

#Digit character class

```python
txt = 'That will be 59 dollers'

x = re.findall(r"\d",txt)

print(x)
```

```python
txt = 'That will be 59 dollers'

x = re.findall(r"\d+",txt)
print(x)
```

```python
txt = "That will be 599 dollers"

x = re.findall(r"\d{2}",txt)
print(x)
```

```python
txt = "That will be 599 dollers 34 dollers and 1 dollers"

x = re.findall(r"\d{2,3}",txt)
print(x)
```

```python
txt ='this is my number 01614-793091'
x = re.findall(r"\d{5}-\d{6}",txt)

print(x)
```

```python
#at the beginning

txt = "123  123 this is my number 01905-672239"
x = re.findall(r"^\d+",txt)

print(x)
```

```python
# at the end

txt = "123  123 this is my number 01905-672239"
x = re.findall(r"\d+$",txt)

print(x)
```

#Wildcard character .

```python
txt = "hello"
x = re.findall(r".",txt)
print(x)
```

```python
txt = "hello"
x = re.findall(r"he..o",txt)
print(x)
```

#Asterics

```python
txt = "rayhan@gmail.com"
x = re.findall(r".*@.*\..*",txt)

print(x)
```

```python
txt = "rayhan@gmail.com  df@gmail.com cat@gmail.com mat@gmail.com dog@yahoo.com"
x = re.findall(r".*@.*\.com$",txt)
print(x)
```
