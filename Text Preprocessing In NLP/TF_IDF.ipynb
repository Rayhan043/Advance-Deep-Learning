```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
corpus = [
    "students watch youtube",
    "youtube watch youtube",
    "students write comments",
    "youtube write comments"
]
```

```python
# dataset make
df = pd.DataFrame({"text" : corpus,"Output" : [1 ,1, 0, 0]})

print(df)
```

```python
# create TF -IDF Vectorizer
tfidf = TfidfVectorizer()
```

```python
X = tfidf.fit_transform(df['text'])
```

```python
#Convert Sparse Matrix to

tfidf_df = pd.DataFrame(
    X.toarray(),
    columns = tfidf.get_feature_names_out()
)

print(tfidf_df)
```

```python
#Print Vocabulary
print(f"Vocabulary : {tfidf.vocabulary_}")
```

```python
#print Feature names
print(f"Feature names : {tfidf.get_feature_names_out()}")
```

```python
#Print IDF Values

print("\nIDF Values:")
for word, idf in zip(tfidf.get_feature_names_out(), tfidf.idf_):
    print(f"{word:10} -> {idf:.4f}")
```

```python
#Transform New Sentence (Inference)
new_text = ["Fuck your Mom"]

new_vector = tfidf.transform(new_text)
print("\nNew Sentence:")
print(new_text[0])

print("\nTF-IDF Vector for New Sentence:")
print(pd.DataFrame(
    new_vector.toarray(),
    columns=tfidf.get_feature_names_out()
))
```

```python
#Example with OOV Word
new_text_oov = ["people watch youtube"]

new_vector_oov = tfidf.transform(new_text_oov)

print('\nSentence with new word')
print(new_text_oov[0])


print("\nTF-IDF vector")
print(pd.DataFrame(
    new_vector_oov.toarray(),
    columns = tfidf.get_feature_names_out()
))
```
