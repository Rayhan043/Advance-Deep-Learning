```python
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd
```

```python
# ==========================================
# 1. Documents (Corpus)
# ==========================================
corpus = [
    "students watch youtube youtube youtube",
    "youtube watch youtube",
    "students write comments comments",
    "youtube write comments"
]
```

```python
#create bang of word  model

cv = CountVectorizer(ngram_range = (2,2))
```

```python
#learn vocavullary and transforn document
X = cv.fit_transform(corpus)
```

```python
cv.get_feature_names_out()
```

```python
# show bag of words matrix

X.toarray()
```

```python
bow_df =pd.DataFrame(
    X.toarray(),
    columns=cv.get_feature_names_out(),
    index=["doc1","doc2","doc3","doc4"]
)

bow_df
```
